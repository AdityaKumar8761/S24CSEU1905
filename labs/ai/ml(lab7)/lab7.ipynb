```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
data = load_breast_cancer().data
target = load_breast_cancer().target
data
```

```python
target
```

```python
data_train, data_test, target_train, target_test = train_test_split(data,target,test_size=0.2,random_state=42,stratify=target)
print("Training data:", data_train.shape)
print("Testing data:", data_test.shape)
```

```python
scaler = StandardScaler()

data_train_scaled = scaler.fit_transform(data_train)
data_test_scaled = scaler.transform(data_test)

```

```python
experiments = [{
        "hidden_layer_sizes": ( 20 , 10),"alpha": 0.0001,"learning_rate_init": 0.001,
    },{
        "hidden_layer_sizes": (50, 100),"alpha": 0.0002,"learning_rate_init": 0.002,
    },{
        "hidden_layer_sizes": (100,50),"alpha": 0.0003,"learning_rate_init": 0.003,
    },{
        "hidden_layer_sizes": (100,50),"alpha": 0.1,"learning_rate_init": 0.004,
    }
]
```

```python
from sklearn.neural_network import MLPClassifier

for i in range(0,4):
    params = experiments[i]

    clf = MLPClassifier(
        hidden_layer_sizes=params["hidden_layer_sizes"],
        alpha=params["alpha"],
        learning_rate_init=params["learning_rate_init"],
        random_state=42,
        max_iter= 300
    ).fit(data_train_scaled, target_train)

    predictions = clf.predict(data_test_scaled[:1])
    clf.predict(data_test_scaled[:5, :])
    train_accuracy = clf.score(data_train_scaled, target_train)
    test_accuracy = clf.score(data_test_scaled, target_test)


    print("traning acuracy", train_accuracy)
    print("test acuracy:", test_accuracy)
```
